**Greedy Best First Search (GBFS).**

Esta función implementa el algoritmo GBFS, en el cual se usa una cola de prioridad basada en los valores de heurística de los nodos, sin considerar el costo acumulado de las acciones. El problema que se usa como prueba es el mostrado en el árbol de búsqueda de la Figura.

```python
import heapq


def gbfs(problema, inicio, meta, h):
	# Cola de prioridad con la heurística del nodo inicial, el nodo inicial y el camino hacia el nodo inicial.
    priorityQueue = [(h[inicio], inicio, [inicio])]
    visitados = set()

    while priorityQueue:
    	# Con heappop se extrae el mínimo valor de heurística, el nodo correspondiente y el camino para llegar a él.
        heuristica, nodo, camino = heapq.heappop(priorityQueue)
        if nodo == meta:
            return camino, visitados
		# Si se cumple esta condición, inicia un nuevo ciclo while.
        if nodo in visitados:
            continue

        visitados.add(nodo)

        for sucesor in problema[nodo]:
            if sucesor not in visitados:
                nuevo_camino = camino + [sucesor]
				# Se inserta el sucesor, su heurística y el camino que lleva a él en la cola de prioridad.
                heapq.heappush(
                    priorityQueue,
                    (h[sucesor], sucesor, nuevo_camino)
                )

    return None
# Se crea un árbol de búsqueda de prueba.
problema = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Se crea un diccionario que asocia nodos con heurísticas.
h = {
    'A': 4,
    'B': 1,
    'C': 2,
    'D': 3,
    'E': 3,
    'F': 2,
    'G': 0
}    
    
camino, visitados = gbfs(problema, 'A', 'G', h)
print('El camino a la meta usando GBFS es:', camino)
print('Los nodos visitados son antes de llegar a la meta son:', visitados)
```

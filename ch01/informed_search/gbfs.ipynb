# 🧭 Greedy Best-First Search (GBFS)

La palabra Greedy dentro del nombre del este algoritmo hace alusión a una `búsqueda codiciosa` basada netamente en los valores de heurísitica y no en los costos de las accciones, lo cual puede conducir a soluciones no óptimas. Este algoritmo utiliza una **cola de prioridad** que organiza los nodos de acuerdo con sus valores de heurística, dando preferencia a aquellos que parecen estar más cerca de alcanzar la meta.

### ⚙️ Funcionamiento del algoritmo

La función de evaluación utilizada por GBFS es:

$
f(n) = h(n),
$

donde $h(n)$ representa la estimación del costo restante desde el nodo $n$ hasta la meta.

Como ya se mencionó, a diferencia de UCS, GBFS **no considera el costo acumulado** de las acciones desde el nodo inicial, $g(n)$. Su decisión se basa exclusivamente en la información proporcionada por la heurística.

### 🧪 Variación del problema con cambios en valores de heurística

Es posible modificar los valores de heurística de los nodos con el fin de observar cómo cambian el orden de expansión y la ruta encontrada por GBFS. En particular, puede comparar los resultados obtenidos al utilizar diferentes heurísticas para identificar situaciones en las que el algoritmo encuentra rápidamente una solución, pero esta no corresponde necesariamente a la de menor costo.


```python
import heapq


def gbfs(problema, inicio, meta, h):
    """
    Algoritmo GBFS con cola de prioridad basada solamente en heurística.
    
    Args:
        problema: árbol o grafo de búsqueda.
        inicio: nodo inicial.
        meta: nodo meta.
        h: diccionario con las heurísticas de cada nodo.
    Returns:
        camino: ruta óptima a la meta.
        visitados: listado de nodos visitados en el proceso de búsqueda.
        None: cuando no hay solución.
    Raises:
        None.
    """
	# Cola de prioridad con la heurística del nodo inicial, el nodo inicial y el camino hacia el nodo inicial.
    priorityQueue = [(h[inicio], inicio, [inicio],0)]
    visitados = set()
    costo_total=0
    while priorityQueue:
    	# Con heappop se extrae el mínimo valor de heurística, el nodo correspondiente y el camino para llegar a él.
        heuristica, nodo, camino, costo_actual = heapq.heappop(priorityQueue)# El costo de cada nodo se extrae solamente para 
        #conocer el costo de la ruta seleccionada. Este no se usa para definir la ruta como si lo hace UCS.
        if nodo == meta:
            return camino, visitados, costo_actual
		# Si se cumple esta condición, inicia un nuevo ciclo while.
        if nodo in visitados:
            continue

        visitados.add(nodo)

        for sucesor, costo_accion in problema[nodo]:
            if sucesor not in visitados:
                nuevo_costo = costo_actual + costo_accion
                nuevo_camino = camino + [sucesor]
				# Se inserta el sucesor, su heurística y el camino que lleva a él en la cola de prioridad.
                heapq.heappush(
                    priorityQueue,
                    (h[sucesor], sucesor, nuevo_camino, nuevo_costo)# El costo de cada nodo se guarda solamente para saber el costo total
                    #para llegar a la meta. Este no se usa para definir la ruta como si se hace en UCS.
                )

    return None
# Usamos el árbol de la Figura 1.
problema = {
    '1': [('2',2), ('6',1),('9',2)],
    '2': [('3',2)],
    '3': [('4',2)],
    '4': [('5',2)],
    '5': [],
    '6': [('7',2)],
    '7': [('8',10),('13',1)],
    '8': [],
    '9': [('10',2)],
    '10': [('11',2)],
    '11': [('12',2)],
    '12': [],
    '13': [('4',1)]
}

# Se crea un diccionario que asocia nodos con heurísticas.
h = {
    '1': 5,
    '2': 3,
    '3': 2,
    '4': 1,
    '5': 0,
    '6': 4,
    '7': 3,
    '8': 1000,
    '9': 30,
    '10': 50,
    '11': 60,
    '12': 100,
    '13': 2,
}    
    
camino, visitados, c_acumulado = gbfs(problema, '1', '5', h)
print('El camino a la meta usando GBFS es:', camino)
print('Los nodos visitados antes de llegar a la meta son:', visitados)
print('El costo acumulado para llegar a la meta es:', c_acumulado)
```

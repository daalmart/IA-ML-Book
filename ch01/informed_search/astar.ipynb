# ⭐ Algoritmo A* (A estrella)

Se presenta a continuación la implementación del algoritmo **A\***. Para ilustrar su funcionamiento y analizar algunas de sus propiedades, se considera un grafo con heurística consistente y un árbol con heurística admisible.

### 🔍 Caso 1: Heurística consistente

En el primer ejemplo se utiliza una heurística consistente. Se busca comprobar que **cada nodo es expandido una sola vez** durante la ejecución del algoritmo. Esto puede verificarse mediante el objeto `Counter` denominado `expansiones`, que almacena el número de veces que cada nodo es analizado (expandido). En ese sentido, el contenido de `expansiones` es {`'A'`: 1, `'B'`: 1, `'C'`: 1} y la ruta óptima $
A \rightarrow C \rightarrow G $, con un costo de 6. Observe que el elemento `expansiones` juega un papel similar al listado de nodos visitados usado en algoritmos tales como DFS o BFS.

### 🎯 Caso 2: Heurística admisible

En el segundo ejemplo se utiliza una heurística admisible. En este caso, se busca comprobar que A* encuentra una **ruta óptima hacia la meta**:

$
A \rightarrow B \rightarrow D \rightarrow F \rightarrow G
$

con un costo total de:

$
C^* = 8.
$

Se propone modificar la heurística del nodo $B$, estableciendo $h(B)=10$. Con este cambio, se obtiene una heurística no admisible y se puede generar una ruta no óptima. Sin embargo, es importante recordar que **una heurística no admisible no implica necesariamente que A* encuentre una solución no óptima**. Lo que sí podemos asegurar es que una heurística admisible garantiza la optimalidad de la solución.

### 🧪 Casos adicionales

Para explorar otros comportamientos del algoritmo, modifique los grafos y las heurísticas utilizando los demás ejemplos planteados en el libro guía.


```python
import heapq
from collections import Counter

def a_star(problema, h, inicio, meta):
    """
    Algoritmo Astar-Cola de prioridad basada en f(n)=g(n)+h(n).
    
    Args:
        problema: árbol o grafo de búsqueda.
        h: diccionario con las heurísticas de los nodos.
        inicio: nodo inicial.
        meta: nodo meta.
    Returns:
        camino: ruta óptima a la meta.
        costo_actual: el costo acumulado hasta la meta.
        expansiones: diccionario con el número de veces que se expande cada nodo.
        costos: diccionario con los costos mónimos de cada nodo
        None: cuando no hay solución.
    Raises:
        None.
    """
    priorityQueue = [(h[inicio], 0, inicio, [inicio])]# Se crea la cola de prioridad con el costo del nodo inicial , el nodo inicial y su ruta
    
    costos = {inicio: 0}# Diccionario para actualizar costos acumulados
    expansiones = Counter()# Permite contar el número de veces que aparece un nodo

    while priorityQueue:
        f_actual, costo_actual, nodo_actual, camino = heapq.heappop(priorityQueue)# Se extrae el costo total, el costo acumulado, el nodo y la ruta hacia él
		
		# Descartar entradas obsoletas
        if costo_actual > costos[nodo_actual]:
            continue

        if nodo_actual == meta:
            return camino, costo_actual, expansiones, costos

        expansiones[nodo_actual] += 1 # Se incrementa cada vez que un nodo es expandido

        for sucesor, costo_accion in problema[nodo_actual]:
            nuevo_costo = costo_actual + costo_accion

            if sucesor not in costos or nuevo_costo < costos[sucesor]:# Se actualizan costos de nodos cuando hay rutas mejores hacia ellos
            
                costos[sucesor] = nuevo_costo
                f_sucesor = nuevo_costo + h[sucesor]

                heapq.heappush(
                    priorityQueue,
                    (f_sucesor, nuevo_costo, sucesor, camino + [sucesor])
                )

    return None

# Creamos un problema de prueba (Figura 1).
problema_caso_1 = {
    'A': [('B', 1), ('C', 3)],
    'B': [('C', 4)],
    'C': [('G', 3)]
}

# Creamos un diccionario de heurísticas para el problema.
h_caso_1 = {
    'A': 4, 
    'B': 5, 
    'C': 3, 
    'G': 0
}
# Llamamos a la función A*
camino, costo, expansiones, costos=a_star(problema_caso_1, h_caso_1, 'A', 'G')
print('Camino hacia la meta:',camino)
print('Costo para acumulado para llegar a la meta:',costo)
print('Nodos y el número de veces que se expandió cada uno:',expansiones)
print('Costos de los nodos:',costos)
```

# 🔎 Búsqueda en profundidad (DFS) y búsqueda en amplitud (BFS)

En esta sección se presenta la implementación de los algoritmos **Depth First Search (DFS)** y **Breadth First Search (BFS)**. El objetivo, además de mostrar su formulación computational en Python, es comparar su desempeño en términos de tiempo de ejecución utilizando diferentes árboles de búsqueda.

### ⚙️ Implementación y medición del tiempo

Cada algoritmo se implementa como una función de Python, que posteriormente se ejecuta \(N\) veces para calcular su **tiempo de ejecución promedio**. Esta repetición es importante porque, en problemas pequeños como el árbol de búsqueda inicial, los tiempos de ejecución pueden ser muy reducidos y, por tanto, sensibles a pequeñas variaciones del entorno de ejecución. Para obtener una medición más representativa, se toman varias muestras y se calcula su promedio:

$
\overline{T} = \frac{1}{N}\sum_{i=1}^{N} T_i
$

donde $T_i$ representa el tiempo de ejecución obtenido en cada repetición.

### 🧪 Variación del problema con aumento de profundidad de árbol

Es posible modificar el problema inicial y explorar cómo cambia el desempeño de ambos algoritmos. Para ello, se puede utilizar la función **`crear_problema()`**, disponible en el paquete `dynamicTree`, que permite generar árboles binarios con diferente profundidad $m$. Esto permite analizar experimentalmente cómo influye el tamaño del espacio de búsqueda en el tiempo de ejecución de DFS y BFS.

```python
#!pip install import-ipynb #Use esta línea solamente la primera vez que ejecute el script.
import import_ipynb #Permite importar otro NoteBook ubicado en la misma ruta.
from resources import dynamicTree #Importamos el NoteBook que genera árboles binarios de m variable.



#Algoritmo DFS-Pila LIFO
def dfs(problema, inicio, meta):
    """
    Algoritmo DFS-Pila LIFO.
    
    Args:
        problema: árbol o grafo de búsqueda.
        inicio: nodo inicial.
        meta: nodo meta.
    Returns:
        camino: ruta óptima a la meta.
        visitados: listado de nodos visitados en el proceso de búsqueda.
        None: cuando no hay solución.
    Raises:
        None.
    """
    LIFO = [(inicio, [inicio])]
    visitados = set()

    while LIFO:#Retorna True si la pila tiene elementos.
        nodo, camino = LIFO.pop()#Retorna el último nodo que entró y el camino que llevó hasta él.
        if nodo == meta:
            return camino, visitados #Si el nodo es la meta, se retorna el camino que llevó a él y se termina el bucle. También se retorna la lista de nodos visitados.

        if nodo not in visitados:
            visitados.add(nodo)

            for sucesor in problema[nodo]:
                if sucesor not in visitados:
                    LIFO.append(
                        (sucesor, camino + [sucesor])#Se almacena cada sucesor y el camino que llevó a él (secuencia de nodos).
                    )

    return None

    
#Algoritmo BFS-Cola FIFO
def bfs(problema, inicio, meta):
    """
    Algoritmo BFS-Cola FIFO.
    
    Args:
        problema: árbol o grafo de búsqueda.
        inicio: nodo inicial.
        meta: nodo meta.
    Returns:
        camino: ruta óptima a la meta.
        visitados: listado de nodos visitados en el proceso de búsqueda.
        None: cuando no hay solución.
    Raises:
        None.
    """
    FIFO = [(inicio, [inicio])]
    visitados = set()

    while FIFO:#Retorna True si la pila tiene elementos.
        nodo, camino = FIFO.pop(0)#Retorna el nodo ubicado en la posición 0 de la lista (el primero que entró) y el camino que llevó hasta él. 
        if nodo == meta:
            return camino, visitados #Si el nodo es la meta, se retorna el camino que llevó a él y se termina el bucle. También se retorna la lista de nodos visitados.

        if nodo not in visitados:
            visitados.add(nodo)

            for sucesor in problema[nodo]:
                if sucesor not in visitados:
                    FIFO.append(
                        (sucesor, camino + [sucesor])#Se almacena cada sucesor y el camino que llevó a él (secuencia de nodos).
                    )

    return None

#Árbol que representa al problema. Usamos el mismo para los dos algoritmos.
problema = {
	'A': ['B', 'C'],
	'B': ['D', 'E'],
	'C': ['F'],
	'D': [],
	'E': [],
	'F': []
}

#La siguiente línea crea un árbol binario de profundidad variable que también puede usarse para probar a los algoritmos. Descoméntela y varíe el valor de m a su gusto.
problema=dynamicTree.crear_problema(m=7)

import time #Se importa time para contar tiempo de ejecución de ambos algoritmos.
N=100000 #Número de ejecuciones para cada algoritmo.

# Conteo para DFS:
inicio_dfs = time.perf_counter()
for _ in range(N):
    cam, visitados_dfs=dfs(problema, '0','10')#Se llama a la función con la información del problema. Se recibe el camino a la meta y los nodos que se visitaron.
fin_dfs = time.perf_counter()
print('El camino a la meta usando DFS es: ',cam)#Imprimimos la ruta que debe seguir el agente según DFS.
print(f"Tiempo de ejecución DFS: {(fin_dfs - inicio_dfs)/N:.8f} segundos")
print("Nodos visitados en DFS: ",visitados_dfs)#Imprimimos los nodos que se visitaron hasta llegar a la meta con DFS.

# Conteo para BFS:
inicio_bfs = time.perf_counter()
for _ in range(N):
    cam, visitados_bfs=bfs(problema, '0','10')#Se llama a la función con la información del problema. Se recibe el camino a la meta y los nodos que se visitaron.
fin_bfs = time.perf_counter()
print('El camino a la meta en BFS es: ',cam)#Imprimimos la ruta que debe seguir el agente según BFS.
print(f"Tiempo de ejecución BFS: {(fin_bfs - inicio_bfs)/N:.8f} segundos")
print("Nodos visitados en BFS: ",visitados_bfs)#Imprimimos los nodos que se visitaron hasta llegar a la meta con BFS.
```

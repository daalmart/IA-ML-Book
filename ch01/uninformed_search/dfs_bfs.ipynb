**DFS y BFS.**

En esta sección se muestra la implementación de los algoritmos DFS y BFS, cada uno como una función Python que posteriormente es llamada N veces para obtener su tiempo de ejecución promedio y poder hacer una comparación. El múltiple llamado de la función de cada algoritmo es necesario porque el problema (árbol de búsqueda) en este caso puede tomar tiempos muy pequeños que son muy suceptibles a pequeñas variaciones, es por eso que se toman varias muestras de tiempo de ejecución y se promedian.

```python
#Algoritmo DFS-Pila LIFO
def dfs(problema, inicio, meta):
    LIFO = [(inicio, [inicio])]
    visitados = set()

    while LIFO:#Retorna True si la pila tiene elementos.
        nodo, camino = LIFO.pop()#Retorna el último nodo que entró y el camino que llevó hasta él.
        if nodo == meta:
            return camino, visitados #Si el nodo es la meta, se retorna el camino que llevó a él y se termina el bucle. También se retorna la lista de nodos visitados.

        if nodo not in visitados:
            visitados.add(nodo)

            for sucesor in problema[nodo]:
                if sucesor not in visitados:
                    LIFO.append(
                        (sucesor, camino + [sucesor])#Se almacena cada sucesor y el camino que llevó a él (secuencia de nodos).
                    )

    return None

    
#Algoritmo BFS-Pila FIFO
def bfs(problema, inicio, meta):
    FIFO = [(inicio, [inicio])]
    visitados = set()

    while FIFO:#Retorna True si la pila tiene elementos.
        nodo, camino = FIFO.pop(0)#Retorna el nodo ubicado en la posición 0 de la lista (el primero que entró) y el camino que llevó hasta él. 
        if nodo == meta:
            return camino, visitados #Si el nodo es la meta, se retorna el camino que llevó a él y se termina el bucle. También se retorna la lista de nodos visitados.

        if nodo not in visitados:
            visitados.add(nodo)

            for sucesor in problema[nodo]:
                if sucesor not in visitados:
                    FIFO.append(
                        (sucesor, camino + [sucesor])#Se almacena cada sucesor y el camino que llevó a él (secuencia de nodos).
                    )

    return None

#Árbol que representa al problema. Usamos el mismo para los dos algoritmos.
problema = {
	'A': ['B', 'C'],
	'B': ['D', 'E'],
	'C': ['F'],
	'D': [],
	'E': [],
	'F': []
}

import time #Se importa time para contar tiempo de ejecución de ambos algoritmos.
N=10000 #Número de ejecuciones para cada algoritmo.

# Conteo para DFS:
inicio_dfs = time.perf_counter()
for _ in range(N):
    cam, visitados_dfs=dfs(problema, 'A','D')#Se llama a la función con la información del problema. Se recibe el camino a la meta y los nodos que se visitaron.
fin_dfs = time.perf_counter()
print('El camino a la meta usando DFS es: ',cam)#Imprimimos la ruta que debe seguir el agente según DFS.
print(f"Tiempo de ejecución DFS: {(fin_dfs - inicio_dfs)/N:.8f} segundos")
print("Nodos visitados en DFS: ",visitados_dfs)#Imprimimos los nodos que se visitaron hasta llegar a la meta.

# Conteo para BFS:
inicio_bfs = time.perf_counter()
for _ in range(N):
    cam, visitados_bfs=bfs(problema, 'A','D')##Se llama a la función con la información del problema. Se recibe el camino a la meta y los nodos que se visitaron.
fin_bfs = time.perf_counter()
print('El camino a la meta en BFS es: ',cam)#Imprimimos la ruta que debe seguir el agente según BFS.
print(f"Tiempo de ejecución: {(fin_bfs - inicio_bfs)/N:.8f} segundos")
print("Nodos visitados en BFS: ",visitados_bfs)#Imprimimos los nodos que se visitaron hasta llegar a la meta.
```

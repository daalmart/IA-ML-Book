```python
def crear_problema(m):
    """
    Crea un árbol binario de profundidad m.
    
    Args:
        m: Es la profundidad deseada del árbol.
    
    Returns:
        Retorna un diccionario que representa al árbol.
    
    Raises:
        None.
    """
    b=2
    total_nodos=0
    tree={}
    for n in range(m+1):
        total_nodos+=b**n
    claves = list(range (int(total_nodos)))
    valores = list(range(total_nodos))
    for i in claves:
        try:
            tree.update({str(claves[i]): [str(valores[i*2+1]),str(valores[i*2+2])]})
        except IndexError:
            tree.update({str(claves[i]): []})
    return tree

#problema=crear_problema(2)
#print(problema['0'])
```

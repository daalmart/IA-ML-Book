```python
import time
def dfs(problema, inicio, meta):
    LIFO = [(inicio, [inicio])]
    visitados = set()

    while LIFO:#Retorna True si la pila tiene elementos.
        nodo, camino = LIFO.pop()#Retorna el último nodo que entró y el camino que llevó hasta él.

        if nodo == meta:
            return camino#Si el nodo es la meta, se retorna el camino que llevó a él y se termina el bucle.

        if nodo not in visitados:
            visitados.add(nodo)

            for sucesor in problema[nodo]:
                if sucesor not in visitados:
                    LIFO.append(
                        (sucesor, camino + [sucesor])#Se almacena cada sucesor y el camino que llevó a él (secuencia de nodos).
                    )

    return None

problema = { 
	'A': ['B', 'C'], 
	'B': ['D', 'E'], 
	'C': ['F'], 
	'D': [], 
	'E': [], 
	'F': []
}
N=10000
inicio_dfs = time.perf_counter()
for _ in range(N):    
    cam=dfs(problema, 'A','F')#Se llama a la función con la información del problema. Se recibe el camino a la meta en la variable cam.
fin_dfs = time.perf_counter()
print('El camino a la meta es: ',cam)#Imprimimos la ruta que debe seguir el agente según DFS.
print(f"Tiempo de ejecución: {(fin_dfs - inicio_dfs)/N:.8f} segundos")
```
